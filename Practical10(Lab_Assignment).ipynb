```python
#Book Database
import pandas as pd

# Creating a dummy csv for demonstration (Ignore if you already have the file)
data = {
    'title': ['Python Basics', 'Data Science 101', 'Deep Learning', 'Java Core', 'AI Ethics'],
    'author': ['John Doe', 'Jane Smith', 'John Doe', 'Alan Turing', 'Jane Smith'],
    'edition': [1, 3, 2, 5, 1],
    'publication_year': [2020, 2018, 2022, 2015, 2023],
    'publishing_house': ['TechPress', 'GlobalEdu', 'TechPress', 'OldBooks', 'GlobalEdu'],
    'price': [250, 450, 800, 150, 600]
}
pd.DataFrame(data).to_csv('books.csv', index=False)

# Main Script
df = pd.read_csv('books.csv')

# a) Print complete report in Tabular form
print("--- Complete Book Report ---")
print(df.to_string())
print("\n")

# b) Print list of available books of a given author
auth_name = "John Doe"
print(f"--- Books by {auth_name} ---")
print(df[df['author'] == auth_name])
print("\n")

# c) Print list of available books of a given publishing house
pub_house = "TechPress"
print(f"--- Books by {pub_house} ---")
print(df[df['publishing_house'] == pub_house])
print("\n")

# d) Print Titles of cheapest & costliest book
cheapest = df.loc[df['price'].idxmin(), 'title']
costliest = df.loc[df['price'].idxmax(), 'title']
print(f"Cheapest Book: {cheapest}")
print(f"Costliest Book: {costliest}\n")

# e) Print the list by sorting based on year of publication
print("--- Books Sorted by Year ---")
print(df.sort_values(by='publication_year'))
```

```python
#State Statistics
import pandas as pd

# Data for 5 states
state_data = {
    'State': ['California', 'Texas', 'Florida', 'New York', 'Alaska'],
    'Area': [163696, 268597, 65758, 54555, 665384],
    'Population': [39237836, 29527941, 21781128, 19835913, 733391]
}

df_states = pd.DataFrame(state_data)

# a) Print complete information
print("--- State Information ---")
print(df_states)
print("\n")

# b) State having largest Area
largest_area_state = df_states.loc[df_states['Area'].idxmax(), 'State']
print(f"State with Largest Area: {largest_area_state}")

# c) State having largest Population
largest_pop_state = df_states.loc[df_states['Population'].idxmax(), 'State']
print(f"State with Largest Population: {largest_pop_state}")

# d) Mechanism to calculate Population Density
df_states['Density'] = df_states['Population'] / df_states['Area']
print("\n--- Data with Population Density ---")
print(df_states)

# e) Determine State with highest population density
highest_density_state = df_states.loc[df_states['Density'].idxmax(), 'State']
print(f"\nState with Highest Population Density: {highest_density_state}")
```
