```python
# Lab Assignment 1: Read a file and write its contents in uppercase to a new file
# --- STEP 1: Create a dummy file for Colab ---
with open("input_text.txt", "w") as f:
    f.write("Hello world! This is a test file for Practical 8.")

# --- STEP 2: Main Logic ---
try:
    # Open the source file for reading
    with open("input_text.txt", "r") as source_file:
        content = source_file.read()

    # Process content to uppercase
    upper_content = content.upper()

    # Write to a new file
    with open("output_upper.txt", "w") as dest_file:
        dest_file.write(upper_content)

    print("Success! 'input_text.txt' has been converted to 'output_upper.txt'.")
    print("-" * 30)
    print("Content of Output File:\n", upper_content)

except FileNotFoundError:
    print("The source file was not found.")
```

```python
#Develop an application using file handling to copy the contents of a python script into another without including the comments. Ask the user about the source and destination file names. Print the content of both the files.
# --- STEP 1: Create a dummy Python script for Colab ---
with open("source_script.py", "w") as f:
    f.write("# This is a header comment\n")
    f.write("print('Hello from the script')\n")
    f.write("x = 10 # This inline comment will remain (common logic)\n")
    f.write("# This is a footer comment")

# --- STEP 2: Main Logic ---
source = input("Enter source file name (e.g., source_script.py): ")
destination = input("Enter destination file name (e.g., cleaned_script.py): ")

try:
    with open(source, "r") as src, open(destination, "w") as dest:
        for line in src:
            # Check if line starts with '#' (ignoring leading whitespace)
            if not line.strip().startswith("#"):
                dest.write(line)

    print("\n--- Operation Complete ---")

    # Printing content of both files
    print(f"\n[Content of Source: {source}]")
    with open(source, "r") as f:
        print(f.read())

    print(f"\n[Content of Destination (No Comments): {destination}]")
    with open(destination, "r") as f:
        print(f.read())

except FileNotFoundError:
    print("Error: One of the files was not found.")
```
