```python
#Diamonds Dataframe Analysis
import pandas as pd

# Creating the sample dataframe from the image
data = {
    'carat': [0.23, 0.21, 0.23, 0.29, 0.31],
    'cut': ['Ideal', 'Premium', 'Good', 'Premium', 'Good'],
    'color': ['E', 'E', 'E', 'I', 'J'],
    'clarity': ['SI2', 'SI1', 'VS1', 'VS2', 'SI2'],
    'depth': [61.5, 59.8, 56.9, 62.4, 63.3],
    'table': [55.0, 61.0, 65.0, 58.0, 58.0],
    'price': [326, 326, 327, 334, 335],
    'x': [3.95, 3.89, 4.05, 4.20, 4.34],
    'y': [3.98, 3.84, 4.07, 4.23, 4.35],
    'z': [2.43, 2.31, 2.31, 2.63, 2.75]
}
df = pd.DataFrame(data)

# i) Calculate the mean price for each cut
mean_price = df.groupby('cut')['price'].mean()
print("--- Mean Price by Cut ---")
print(mean_price)

# ii) Print count, minimum, and maximum price for each cut
# Using .agg() allows us to get multiple statistics at once
cut_stats = df.groupby('cut')['price'].agg(['count', 'min', 'max'])
print("\n--- Price Statistics (Count, Min, Max) by Cut ---")
print(cut_stats)

# iii) Calculate and print average value of x, y, and z separately
avg_xyz = df[['x', 'y', 'z']].mean()
print("\n--- Average of x, y, and z ---")
print(avg_xyz)
```

```python
#Employee Report Program
import pandas as pd
import os

file_name = 'employee.xlsx'
if not os.path.exists(file_name):
    dummy_data = {
        'Employee ID': [101, 102, 103, 104],
        'Employee Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Department': ['Automotive', 'Banking', 'Automotive', 'IT'],
        'Designation': ['Senior Developer', 'Manager', 'Junior Developer', 'Tester']
    }
    pd.DataFrame(dummy_data).to_excel(file_name, index=False)
    print(f"Created a temporary {file_name} for demonstration.\n")

# --- STEP 2: The actual Assignment Logic ---
try:
    # Read the file
    emp_df = pd.read_excel(file_name)

    # a) Print the list of employees working for "Automotive" domain
    # Use .str.strip() to handle accidental spaces in Excel cells
    automotive_emps = emp_df[emp_df['Department'].str.strip() == 'Automotive']
    print("--- Employees in Automotive Domain ---")
    print(automotive_emps)

    # b) Print details of an employee with employee ID given by an end user
    # We use .astype(str) to ensure the comparison works regardless of formatting
    search_id = input("\nEnter Employee ID to search (e.g., 101): ").strip()
    employee_details = emp_df[emp_df['Employee ID'].astype(str) == search_id]

    if not employee_details.empty:
        print("\n--- Employee Details ---")
        print(employee_details)
    else:
        print(f"No employee found with ID: {search_id}")

    # d) Print the list of all the Developers of Infosys
    # .str.contains() is great for finding 'Developer' within 'Senior Developer'
    developers = emp_df[emp_df['Designation'].str.contains('Developer', case=False, na=False)]
    print("\n--- List of Developers ---")
    print(developers)

except Exception as e:
    print(f"An error occurred: {e}")
    print("Tip: Make sure you have installed openpyxl: pip install openpyxl")
```
